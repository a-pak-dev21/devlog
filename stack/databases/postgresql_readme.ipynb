### **UNION**
    used between 2 queries to return result of 2 or more queries as a single-set(1 column) result.
    Return only unique values

### **UNION ALL**
    works same way as *UNION* but also return all duplicates

### **INSERT**
    to insert new row into a table we use command INSERT INTO
    syntax: INSERT INTO <table_name> (column_name1, column_name2, column_name3, ...)
            VALUES (value_for_col1, value_for_col2, value_for_col3), (val2_for_col1, ...), (), ...

### **UPDATE**
    update values by conditions 
    syntax: UPDATE <table_name>
            SET column1 = value1, column2 = value2, ...
            WHERE <condition>

### **DELETE**
    DELETE rows by conditions
    syntax: DELETE FROM <table_name>
            WHERE <condition>

### **EXTRACT**
    EXTRACT(<timestamp> FROM datetime) return variable in <timestamp> from datetime variable
    where timestamp might be time or part of date - YEAR, MONTH, WEEK, DAY
    
### **CONCAT**
    CONCAT(string1/var1, string2/var2, string3/var3, ...) or other syntax is 
    string1/var1 || string2/var2 || string3/var3 || ...

### **ROLLUP**
    used inside GROUP BY to do same as normal group by by few columns, but also return subresult for subgroup and total result of all
    for agg function


### **WINDOW functions**
    All window functions using word *OVER*. WOrk similar to a GROUP BY, the difference is that it return value from your function
     from specific range in a specific order as you wish and main for each element. So it's not grouping them together
    General syntax:
        <funciton_name> OVER (PARTITION BY <group_keys> ORDER BY<group_keys>, other arguements)


## DIFFERENT DATA TYPES

### Character types
- char(n) - stores up to a max number of n characters(если строка меньше значения n то заполнит пробелами)
- varchar(n) - do same as char(n) but not autofill with whitespaces till n symbols just flexible length
- varchar / text  - store any length of characters

### Numeric types
- Serial - whole numbers that also **auto-increment**. Always used for *columns ids* (3types: smallserial, serial, bigserial)
- Integer - whole numbers only, always when we don't need decimals ( smallint, int, bigint)
- floats - numbers with decimals:
    - Decimal
    - Numeric
    - Real
    - Double precision
    - Float - same as double
  
### Boolean types
- True, 1, t, y, yes, on
- False, 0, f, n, no, off
- Null


### date / time
- !!! No matter what format enter the date, it will be transformed into format: "YYYY-MM-DD" 
- '1:30:30 PM':: TIME WITHOUT TIME ZONE -> 13:30:30
- 01:30 PM EST/PST/UTC 

## Create type
    We can create new types inside our database by defining them as:
    syntax: CREATE TYPE <type_name> AS <data_type(for example enum)> ('val1', 'val2', 'val3',...)
        In this case enum is enumerated set of values from which our new data type can be chosen
